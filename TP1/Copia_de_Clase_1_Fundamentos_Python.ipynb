## 🧮 Variables

Una **variable** es un espacio donde guardamos información.

Nos permite reutilizar datos más adelante en el código.

```python
nombre = "Ana"
edad = 35

print(nombre)
print(edad)
```

## 📚 Listas

Una **lista** permite almacenar múltiples valores dentro de una misma variable.

Esto es útil cuando trabajamos con colecciones de elementos.

```python
centros = ["Clínica Central", "Hospital Norte", "Clínica Sur"]

print(centros)
```

```python
centros[0]
```

```python
lista2 = [1, 10, "perro"]
print (lista2)
```

## 🧠 Diccionarios

Un **diccionario** permite almacenar información utilizando pares **clave : valor**.

Es una forma muy útil de organizar datos relacionados.

```python
paciente = {
    "nombre": "Ana",
    "edad": 35,
    "ciudad": "Santiago"
}

print(paciente)
```

```python
paciente["edad"]
```

```python
paciente = {
    "nombre": ["Ana","Luis","Camila"],
    "edad": [35, 40, 28],
    "ciudad": ["Santiago", "Valparaiso", "Concepción"]
}

print(paciente)
```

```python
import	pandas	as	pd
df = pd.DataFrame(paciente)
df
```

## 🧩 Funciones

Una **función** es un bloque de código que realiza una tarea específica.

Permite reutilizar lógica sin tener que escribir el mismo código muchas veces.

```python
def saludar(nombre):
    return f"Hola {nombre}, bienvenido al análisis de datos"

saludar("Carlos")
```

```python
def saludar(nombre):
    print (nombre)

saludar("Carlos")
```

## 📊 DataFrames

Cuando trabajamos en análisis de datos, normalmente utilizamos **tablas**.

En Python estas tablas se llaman **DataFrames** y se manejan principalmente con la librería **Pandas**.

```python
import pandas as pd
```

```python
datos = {
    "nombre": ["Ana", "Luis", "Camila"],
    "edad": [25, 32, 29],
    "ciudad": ["Santiago", "Valparaiso", "Concepción"]
}

df = pd.DataFrame(datos)

df
```
